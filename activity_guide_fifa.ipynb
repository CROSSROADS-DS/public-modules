# FIFA Players: Investigating Skill, Age, and Cost

## Your mission

Imagine that **you are managing a FIFA team on a limited budget**.

**Your goal is to develop an evidence-based strategy for choosing the type of player you spend money on.** You cannot simply buy the highest-rated players available; you must work out what kind of player gives them the most for the money.

Your work will produce a strategy you can defend with evidence, along with a clear account of what your team would be good at and what it would struggle with.

## Begin with questions

Start by developing one or more questions that could help you spend a limited budget well. Your questions might involve overall rating, age, individual skills, transfer value, weekly wage, or relationships among these (or other) attributes.

After meeting the dataset, you will decide which questions it can support and refine, replace, or add questions as needed.

## Shared investigation

The class will follow one selected investigation path together:

> **What separates the highest-rated FIFA players from the rest, and how do older players differ from younger ones?**

You will first examine what a high rating is actually made of, comparing the skill profiles of highly rated and lower-rated players. You will then bring in age, discover why the obvious comparison is misleading, and build a fairer one. Finally you will compare what different types of player cost. At each step, the goal is to use evidence to answer a question, not simply to produce a visualization.

## How you will work

The notebook provides short Python helper-function calls. Before each call, the notebook explains its purpose. After it runs, you will examine the output, record observations, cite evidence, and consider what the result does and does not establish.

At three points you will be asked to write down a prediction *before* running the code. Do this honestly, including when you turn out to be wrong. Noticing the gap between what you expected and what the data show is part of the method.

The guided activity prepares you to adjust helper-function options during further exploration.

## Concluding the analysis

After completing the analysis, you will be asked to:

- State the type of player you would spend your budget on, and why
- Support the strategy with at least three pieces of evidence from the investigation
- State what your team's strengths and weaknesses would be as a result
- State the limitations of the analysis and what additional data you would want before committing the budget

This will help you synthesize your results and turn them into a decision you can defend. There is no single correct strategy. What separates a strong answer from a weak one is whether the trade-off is named and supported by evidence.

An optional independent extension allows you to return to the questions you developed at the beginning and investigate one of your own choosing, using the same helper functions with different settings. Explain your choice and present your expectations before analyzing the data, then revisit them afterward to see whether they held.
