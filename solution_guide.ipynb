# Solution guide - FIFA Players: Investigating Skill, Age, and Cost

This guide summarizes expected findings and discussion points. Student responses may differ when they are supported by relevant evidence and acknowledge limitations. The figures below are the values students will see in the notebook output.

## Dataset evaluation

The data support questions about what distinguishes highly rated players, how age relates to skill and to price, and what different types of player cost. Questions involving nationality, preferred foot, or position could also be addressed, though they are not covered in the shared investigation.

The data are not sufficient to address questions such as *which attributes win matches?* The file records how the game rates players, not what those players did in a match, so no claim about performance can be supported. Answering that question would require match outcomes linked to the players involved.

Two other common questions fall outside the data. *How does a player change as he ages?* cannot be answered, because the file is a single snapshot of one season rather than the same players followed over time.

Students may also reasonably note that contract length, injury history, squad chemistry, and league quality are absent, and that every value is assigned by the game rather than observed in a real transfer market.

## Shared investigation

### What information in these rows seems most relevant?

The columns most relevant to a budget decision are `overall_rating`, `age`, `value_euro`, and `wage_euro`, together with the twenty-nine skill attributes describing what a player is actually good at. Strong responses connect a column to one of their own questions rather than simply listing columns.

### Which questions can this dataset answer?

Answerable: what separates highly rated players from the rest, how skill profiles differ by age, and what each type of player costs. Not answerable: which attributes produce wins, how an individual player changes over a career, and anything about goalkeepers. Match-level performance data and several seasons of the same players would be needed.

### Refine your questions

Expect broad questions such as "who are the best players?" to be narrowed into answerable ones such as "what type of player gives the most skill per euro?" What matters is that the revision is driven by what the columns can actually support.

### What is the distribution of overall rating?

The distribution is roughly symmetrical and centred in the middle of its range: mean 66.5, median 66, standard deviation 6.8, running from 47 to 94. Highly rated players are rare. Only 3.0% of players are rated 80 or above and 0.6% are rated 85 or above, while about two thirds sit between 60 and 72.

### Prediction: which skills separate the best players? Which attributes differ most between highly rated and lower-rated players?

The largest differences are all mental or technical: composure (+22.3), reactions (+20.8), ball control (+19.7), vision (+19.4), and short passing (+19.3). The smallest are physical: balance (+3.0), acceleration (+5.1), sprint speed (+5.5), and jumping (+6.7).

On average, highly rated player have much higher composure, reactions, and ball control ratings than lower-rated players. There is much less of a difference in balance, acceleration, and sprint speed ratings between the two player groups. The game devlopers seems to value the mental attributes of players more than the physical ones when it comes to differientating players abilities.

### What is the distribution of age?

Age peaks in the early twenties and is right-skewed: mean 25.4, median 25, range 17 to 40. The two tails differ sharply. The young end stops abruptly at 17, because younger players are not included, while the old end thins gradually across a long tail of 1,162 players aged 33 or over. This is unlike the rating distribution, which falls away symmetrically on both sides. The asymmetry is a first hint that players leave the dataset as they age.

### Prediction: how is age related to rating? How is age related to overall rating?

Average rating climbs steeply from 55.7 at age 17 to about 69 by age 27, then stops changing: 69.6 at 29, 69.9 at 31, 70.0 at 34. The line never falls.

The chart cannot be read as a picture of individual careers because it compares different players at each age rather than following the same players over time. Players who declined were released, retired, or dropped out of the leagues the game covers, so they are absent from the file. The only older players visible are the ones who were still good enough to remain, which is why the line stays flat. Answering the intended question would require longitudinal data: the same players tracked across several seasons, including those who leave.

### Are the player evenly distributed across the age/rating groups?

No, sparse cells sit at opposite corners: 293 players rated 75 and over who are 23 and under, and 695 players rated 64 and under who are 29 and over. Highly rated young players and lowly rated old players are both uncommon, so those two cells rest on the least evidence.

### Does our age normalization work, are cells within a row similar, why might a particular group behave differently from the others?

The "64 and under" row stands out. Average rating across it runs 58.6, 61.6, 62.0, a spread of 3.4 rating points, while the other three rows hold steady to within a point (65 to 69: 66.6, 67.0, 67.1; 70 to 74: 71.6, 71.8, 71.8; 75 and over: 77.4, 78.1, 78.3).

The cause is that this band is much wider than the others, covering ratings 47 to 64, and young players cluster near its bottom. Comparing across that row therefore mixes an age difference with a rating difference, so results from it should carry less weight than results from the narrower bands.

### Prediction: do older and younger players have the same skills? Do older players have different skill sets?

Among players rated 70 to 74, the youngest group is markedly faster: acceleration 74.2 against 64.5 (a gap of 9.7), sprint speed 74.0 against 64.5 (9.5), agility 71.8 against 65.9 (5.9), dribbling 69.0 against 63.5 (5.5). The oldest group leads on aggression (68.1 against 60.8), interceptions (58.3 against 51.9), strength (71.7 against 65.5), free-kick accuracy (54.8 against 48.6), heading accuracy (64.2 against 58.8), and composure (68.7 against 65.7).

The key observation is that the bars point in both directions. This is not one group being better; it is two groups being good at different things. Note also that the skills favouring younger players are largely the ones that barely separated rating groups in section 5, while composure and reactions, which separated them most, favour older players.

### Does the pattern hold in a second rating group?

Yes. In the 65 to 69 group the same attributes appear at the same ends, with similar magnitudes: acceleration -8.4 and sprint speed -8.0 against aggression +8.6, interceptions +7.4, and strength +7.1.

Repeating the comparison matters because a pattern found once could be a quirk of the particular group examined. When the same result appears among a separate set of players, chance and group-specific artefacts become less plausible explanations.

### At the same rating, how much more do younger players cost?

Roughly twice as much. In the 70 to 74 group the median value is EUR 3,900,000 for players 23 and under against EUR 1,900,000 for players 29 and over, a ratio of about 2.1. In the 65 to 69 group it is EUR 1,100,000 against EUR 575,000, about 1.9; in the 75 and over group, EUR 12,000,000 against EUR 7,000,000, about 1.7.

The "64 and under" row does not follow the pattern, which is consistent with the warning raised in section 8 about that row.

### How do value and wage behave differently?

Transfer value falls steeply with age at a fixed rating, but weekly wage stays essentially flat: EUR 11,000, EUR 11,000, EUR 10,000 across the 70 to 74 row, and EUR 4,000 in all three cells of the 65 to 69 row.

The two numbers pay for different things. Wage compensates a player for what he contributes now, and since the players being compared are rated the same, they are paid about the same. Transfer value is the one-time price of the asset itself, reflecting how many seasons remain and what the player could later be sold for. Buying an older player therefore produces a large saving once, not a saving every week.

### What does the extra money buy, and what does it give up?

Following the 70 to 74 row, the additional EUR 2,000,000 buys sprint speed of 74.0 rather than 64.5, and gives up interceptions of 51.9 rather than 58.3. Students should describe the exchange in both directions rather than treating the cheaper player as simply worse.

Strong responses also register the caution stated in the notebook: age changes along the row as well, so these grids cannot separate paying for pace from paying for youth. Holding age constant as well as rating would be needed to tell the two apart.

## Result summation and suggested strategies

Main findings, though students may highlight others:

- Highly rated players are separated from the rest mainly by mental and technical attributes; pace barely distinguishes them at all.
- Average rating never declines with age in this dataset, because declining players have already left it.
- Among players rated the same, older players are substantially slower but stronger, more aggressive, better in the air, and more composed.
- At the same rating, younger players cost roughly twice as much to buy, while weekly wages are about the same.

Two defensible strategies, both supportable from the same evidence:

- **Buy older players at a given rating.** Roughly twice the squad for the money. Strengths: physical duels, aerial play, defensive positioning, composure, set pieces. Weaknesses: a slow team, and a squad that loses resale value quickly.
- **Pay the premium for younger players.** Strengths: pace, and a squad that holds its value for later sale. Weaknesses: fewer players for the budget, and weaker in physical and positional attributes.

Limitations students should raise: the ratings are assigned by the game rather than measured; nothing in the data indicates which attributes actually win matches; and the age comparison covers only the players who lasted in the game long enough to appear in it.

## Acceptable variation in responses

Students may emphasize different attributes, rating groups, or strategies. Strong responses should:

- cite visible or tabular evidence, including specific values;
- use cautious language such as "suggests," "is consistent with," or "may";
- avoid treating interpretations as recorded facts;
- describe what a strategy gives up as well as what it gains;
- connect the finding to the mission of building a team on a limited budget; and
- identify at least one limitation that matters for the claim.
