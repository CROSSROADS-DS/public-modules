### 2. Data Evaluation

The displayed rows provide a first look at the data. Each row represents one movie. Note each element of the data, and read the summary below of the different types of data.

Data in this file can be **text**, a **number**, or **categorical** (an option from a few different choices). There are far more types of data in the world of data science, but these are the types we're using today.

- **name** (text): The name of the movie
- **rating** (categorical): The parental rating, such as PG, R, or PG-13
- **genre** (categorical): The genre of the movie
- **year** (number): The year it was released
- **score** (number): Average score on iMDB, out of 10
- **votes** (number): The number of ratings on iMDB
- **director** (text): The director
- **writer** (text): The writer
- **star** (text): The top-billed actor
- **country** (categorical): The country it was produced in
- **budget** (number): The budget, in dollars. Not adjusted for inflation.
- **gross** (number): The total revenue of the movie, including box office and home media
- **company** (categorical): The company that produced the movie
- **runtime** (number): The runtime, in minutes
- **release_month** (number): The month the movie was released in
- **release_day** (number): The day the movie was released on

### 6. Removing Small Categories

#### Purpose of the Code

Because small genres are impossible to make a broad conclusion about, we are going to remove any genre with less than 30 movies. The `without_small_categories` will remove small genres, and save the new dataset as `data_no_small_categories`. Setting `limit=30` indicates that this is the limit of a "small" genre. Setting `by="genre"` makes sure that the categories are arranged by genre.

We will then show the distribution of this filtered dataset. We can use the same tool to show distributions as before, `plot_number_of_movies_by_genre`. Note how this time, we use `data_no_small_categories` instead of `data`.

```python
data_no_small_categories = without_small_categories(data, limit=30, by="genre")
plot_number_of_movies_by_genre(data_no_small_categories)
```

### 8. Examining Genres' Stats

#### Purpose of the Code

To look at the critical scores of a genre in the last few years, the function `show_recent_summary_details` is used.
```python
show_recent_summary_details(data_no_small_categories, genres=("Comedy", "Drama"), property="score", years=10)
```

From these summary statistics, one can notice that drama movies get reviewed a bit better than comedy movies. Both get rated around the 6.5 range on average, but drama movies have a **median** of 6.7 and action movies only have a median of 6.3. Remember, a median is like an average but better for when big movies bring it up or down! The larger gap when using median means that there may be some highly-rated comedies, but drama is more consistent at producing highly-rated movies.

#### Transition
