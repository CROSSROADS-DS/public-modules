# Movies: Investigating Patterns, Types of Data, and Distribution of Data

## Helper functions

These functions are used in the interactive activity notebook to run code and produce results. This file defines all functions and provides example usages where necessary.


**Function** *load_movie_data()*

Will load in the movie dataset and print out the first five rows. **See Part 1 of the activity notebook.**

**Function** *show_dataset_scope(X)*

Will produce and display summary statistics for dataset *X*, including number of movies by genre and country. **See Part 3.**

**Function** *plot_number_of_movies_by_genre(X)*

Will plot the number of movies by genre in dataset *X* as a barplot. **See Part 4.**

**Function** *show_movies_in_genre(X, genre=Y)*
    
Will display in tabular form the movies for genre *Y* in dataset *X*. **See Part 5.**

**Function** *without_small_categories(X, limit=Z, by=Y)*

Will remove from dataset *X*, small categories within variable *Y* with *Z* or fewer items in the dataset. **See Part 6.**

**Function** *show_recent_summary_details(X, genres=C, property=B, years=A, category_property="Y")*

Will display summary statistics for the property *B* (e.g, ratings) for specified genres *C* over a specified time period in years *A* in dataset *X*. **See Part 8.**

**Function** *show_plot_over_time(X, genres=C, property=B, category_property=Y, time_property=A)*

Will produce a line graph of specified genres *C* over a specified number of years *A*, for a given property *B* (e.g., gross income) in dataset *X*. **See Part 9.**

**Function** *compare_categories(X, size=S, category=C, compare=, compared_by=..., chart_by=...)*

A function that produces all of the previous function output, wrapped into a single function. Inputs are the dataset *X*, the minimum size *S* of the category *C* (e.g., S=30 films for *category=genre*), the specific fields to compare (e.g., *compare=('action','comedy')* for *category=genre*) and the numeric field to compare them using compared_by (e.g., *compared_by="score"*) and how the data should be charted using chart_by (e.g., *chart_by ="Year"*). **See Part 13.**
